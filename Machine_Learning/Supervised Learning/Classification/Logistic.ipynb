```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
#create dataset
X,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=56)
```

```python
#visualize the dataset
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y)
plt.title("Scatter plot of the dataset")
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
model.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score=accuracy_score(y_test,y_pred)
print("Accuracy:",score)
print("Classification Report:\n",classification_report(y_test,y_pred))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

Hyperparameters tuning


```python
logistic_model=LogisticRegression()
penalty=['l1', 'l2', 'elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
cv=StratifiedKFold()
grid_search=GridSearchCV(estimator=logistic_model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
```

```python
grid_search
```

```python
grid_search.fit(X_train,y_train)
```

```python
grid_search.best_params_
```

```python
y_pred=grid_search.predict(X_test)
```

```python
score=accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

## Randomized SearchCV


```python
from sklearn.model_selection import RandomizedSearchCV
random_search=RandomizedSearchCV(estimator=logistic_model,param_distributions=params,scoring='accuracy',cv=5,n_jobs=-1)
random_search.fit(X_train,y_train)
```

```python
random_search.best_params_
```

```python
random_search.best_score_
```

```python
y_pred=random_search.predict(X_test)
```

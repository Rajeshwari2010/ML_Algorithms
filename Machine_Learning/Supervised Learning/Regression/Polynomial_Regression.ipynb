POLYNOMIAL REGRESSION

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X=6*np.random.rand(100,1)-3
Y=0.5* X**2 + 1.5*X + 2 + np.random.rand(100,1)
#quadratic equation used y= ax^2 + bx + c
#where a=0.5, b=1.5 and c=2

```

```python
plt.scatter(X,Y,color='green')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot of X vs Y')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
y_pred=model.predict(X_test)
score=r2_score(Y_test,y_pred)
```

```python
score
```

```python
plt.plot(X_train,model.predict(X_train),color='blue',label='Predicted Data')
plt.scatter(X_train,Y_train,color='green',label='Training Data')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Training Data vs Predicted Data')
plt.legend()
plt.show()
```

```python
#applying polynomial regression
from sklearn.preprocessing import PolynomialFeatures

Poly=PolynomialFeatures(degree=2,include_bias=True)
```

```python
X_train_poly=Poly.fit_transform(X_train)
X_test_poly=Poly.transform(X_test)
```

```python
X_train_poly
```

```python
from sklearn.metrics import mean_squared_error, r2_score
regression=LinearRegression()
regression.fit(X_train_poly,Y_train)
y_pred_poly=regression.predict(X_test_poly)
score_poly=r2_score(Y_test,y_pred_poly)
score_poly
```

```python
#visualization of polynomial regression
plt.scatter(X_train,Y_train,color='green',label='Training Data')
plt.scatter(X_train,regression.predict(X_train_poly),color='blue',label='PolynomialRegression Line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Training Data vs Polynomial Regression Line')
plt.legend()
plt.show()
```

```python
#new data point prediction

new_data=np.linspace(-3,3,200).reshape(200,1)

#np.linspace(-3, 3, 200) creates 200 evenly spaced numbers between -3 and 3 
# inclusive.Calling .reshape(200, 1) turns that 1D array of shape 
# (200,) into a 2D column array with shape (200, 1), which is often required by scikit-learn models 
# expecting a feature matrix (n_samples, n_features) rather than a flat vector.
new_data_poly=Poly.transform(new_data)
new_prediction=regression.predict(new_data_poly)
new_prediction
```

```python
#visualization of polynomial regression
plt.scatter(X_train,Y_train,color='green',label='Training Data')
plt.scatter(X_train,regression.predict(X_train_poly),color='blue',label='PolynomialRegression Line')
plt.plot(new_data,new_prediction,color='red',label='New Data Point')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Training Data vs Polynomial Regression Line with New Data Point')
plt.legend()
plt.show()
```

Pipeline in Polynomial Regresssion
A pipeline is a way to chain multiple steps (transformations + model) into one clean workflow.

📌 Why do we need a pipeline?

In polynomial regression, we don’t directly fit X.

We must:
Convert 
𝑥→𝑥,𝑥^2,x^3...
Then apply linear regression

👉 Pipeline automates this sequence


Without Pipeline (Problem)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, Y)


```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import numpy as np

# Data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([1, 4, 9, 16, 25])  # quadratic relationship

# Create pipeline
model = Pipeline([
    ("poly", PolynomialFeatures(degree=2)),
    ("linear", LinearRegression())
])

# Train
model.fit(X, Y)

# Predict
print(model.predict([[6]]))
```

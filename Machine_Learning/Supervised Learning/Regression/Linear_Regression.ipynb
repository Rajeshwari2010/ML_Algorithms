```python
# Step 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 2: Create dataset (X = experience, Y = salary)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)   # Independent variable
Y = np.array([30000, 35000, 50000, 60000, 70000])  # Dependent variable

# Step 3: Create model
model = LinearRegression()

# Step 4: Train model
model.fit(X, Y)

# Step 5: Make predictions
predicted = model.predict(X)

# Step 6: Print results
print("Intercept (β0):", model.intercept_)
print("Slope (β1):", model.coef_[0])

# Step 7: Predict new value
new_exp = np.array([[6]])
print("Predicted salary for 6 years experience:", model.predict(new_exp)[0])

# Step 8: Visualization
plt.scatter(X, Y, label="Actual Data")
plt.plot(X, predicted, color='red', label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Step 1: Dataset
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
Y = np.array([5, 7, 9, 12, 18, 25])  # slightly non-linear data

# Step 2: Linear Regression (degree 1)
linear_model = LinearRegression()
linear_model.fit(X, Y)
y_linear = linear_model.predict(X)

# Step 3: Polynomial Regression (degree 2)
poly2 = PolynomialFeatures(degree=2)
X_poly2 = poly2.fit_transform(X)

model2 = LinearRegression()
model2.fit(X_poly2, Y)
y_poly2 = model2.predict(X_poly2)

# Step 4: Polynomial Regression (degree 3)
poly3 = PolynomialFeatures(degree=3)
X_poly3 = poly3.fit_transform(X)

model3 = LinearRegression()
model3.fit(X_poly3, Y)
y_poly3 = model3.predict(X_poly3)

# Step 5: Plot everything
plt.scatter(X, Y, label="Actual Data")

plt.plot(X, y_linear, label="Linear Fit (deg=1)")
plt.plot(X, y_poly2, label="Polynomial Fit (deg=2)")
plt.plot(X, y_poly3, label="Polynomial Fit (deg=3)")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Multiple Best Fit Lines")
plt.legend()
plt.show()
```

```python
#Magic Methods -double undersore methods
'''like __init__ ,__str__,__dir__,__hash__,__len__,__module__'''
```

```python

class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self): #return string representation orignally now we override it
        return f"{self.name} is {self.age} years old"    

p1=person("Raj",23)        
print(p1)
```

```python
#operator overloading using magic methods
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
        

    def __add__(self,other):
        return vector(self.x +other.x,self.y+other.y)
    

    def __sub__(self,other):
        return vector(self.x-other.x,self.y-other.y)
    
    def __mul__(self,other):
        return vector(self.x*other,self.y*other)

    def __eq__(self,other):
        return vector(self.x == other.x,self.y == other.y)
v1=vector(2,4)
v2=vector(3,4)
print(v1+v2)  
```

```python
#custom exception- raise and throw

class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year=int(input("Enter date of birth"))
age=2025-year

try:
    if age<=30 and age >=20:
        print("The age is valid")

    else:
        raise dobException 

except dobException:
    print("Age is not valid")      
```

```python
#generator
def square(n):
    for i in range(n):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
#Practical example:Reading large file
#process one line at a time without loading inn memory

def read_large_file(file_path):
    with open(file_path,"r") as file:
        for line in file:
            yield(line)
```

```python
file_path="example.txt"
for line in read_large_file(file_path):
    print(line.strip())
```

```python
#Closure funtion inside a function

def main_welcome():
    msg="welcome!"

    def sub_welcome():
        print("Welcome to the course")
        print(msg)
        print("Im learning omething new")

    return sub_welcome()    
```

```python
#decorator
def main_welcome(func):
    msg="welcome!"

    def sub_welcome():
        print("Welcome to the course")
        func()
        print("Im learning omething new")

    return sub_welcome()   
```

```python
def course_introduction():
    print("This is course introduction")
```

```python
main_welcome(print)
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome #course will be called inside this as a parameter(similar to above cell)
def course_introduction():
    print("This is course introduction")
```

```python
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper

@my_decorator
def greet():
    print("Hello!")

greet()
```

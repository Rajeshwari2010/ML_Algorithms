```python
import sqlite3
```

```python
#connect to sqlite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
#create table
cursor.execute(
    ''' Create Table if not exists Employees(
    id Integer Primary Key,
    Name Varchar Not Null,
    Age Integer,
    Department Varchar
    )'''
)

##commit the changes
connection.commit()
```

```python
cursor.execute(
    ''' Select * from Employees'''
)
```

```python
##Insert dta in table

cursor.execute(''' 
Insert Into Employees(id,Name,Age,Department)
               Values
               (1,"Krish",20,"HR"),
               (2,"Raj",23,"Finance"),
               (3,"Rajeshwari",22,"HR"),
               (4,"Rutu",22,"IT"),
               (5,"Sakshi",23,"IT")

'''
)

connection.commit()
```

```python
#query data from sqlite3
cursor.execute('Select * From Employees')
rows=cursor.fetchall()


#print the queried data
for row in rows:
    print(row)
```

```python
#update dta
cursor.execute(''' 
Update employees
               Set Age=25
               Where Name="Krish"
'''

)

connection.commit()
```

```python
#delete data
cursor.execute('''
Delete From Employees
Where Name="Raj"'''

)

connection.commit()
```

```python
#close connection
connection.close()

#To open connection run connection=sqlite3.conect and cursor again
```

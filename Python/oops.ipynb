```python
class Bank_account:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} deposited. New balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
            print(f"{amount} withdrawn. Remaining Balance is {self.balance}")

bk=Bank_account("Rajeshwari",100000)
bk.deposit(500000)  
bk.withdraw(20000)  
bk.withdraw(1298218937)        
```

```python
#Multiple Inheritance
class Animal:
    def eat(self):
        print("This animal is eating")

    def sleep(self):
        print("This animal is sleeping")

class Flyable:
    def fly(self):
        print("This animal is flying")


class Bird(Animal,Flyable):
    def chirp(self):
        print("The bird is chirping")


b1=Bird()
b1.eat()
b1.fly()
b1.chirp()                            
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks    # private variable

    # Getter method
    def get_marks(self):
        return self.__marks

    # Setter method
    def set_marks(self, marks):
        if 0 <= marks <= 100:   # validation
            self.__marks = marks
        else:
            print("Invalid marks")
```

```python
s = Student("Raj", 85)

print(s.get_marks())     # Getting marks

s.set_marks(95)          # Setting new value
print(s.get_marks())

s.set_marks(150)         # Fails validation
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass   # method has no implementation (hidden details)

    def display(self):
        print("This is a shape")   # normal method


# Child class implementing abstract method
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth


# Another child class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


# Using abstraction
r = Rectangle(10, 20)
c = Circle(5)

print("Rectangle Area:", r.area())
print("Circle Area:", c.area())
```

```python
import numpy as np
```

```python
marks_list=[43,53,23,45,67,89,90,34,56,78,99,100,23,45,67,89,90]
min,Q1,median,Q3,max=np.quantile(marks_list, [0, 0.25, 0.5, 0.75, 1.0])
```

```python
min,Q1,median,Q3,max
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR

lower_limit,upper_limit
```
